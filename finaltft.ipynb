```python
# STEP 5.16.3: Compare target with actual next-week demand

df_sorted["actual_next_week_demand"] = (
    df_sorted.groupby("product_id")["units_sold"].shift(-1)
)

valid_rows = df_sorted["actual_next_week_demand"].notna()

comparison = df_sorted.loc[
    valid_rows,
    ["next_week_demand_target_units", "actual_next_week_demand"]
].copy()

comparison["difference"] = (
    comparison["next_week_demand_target_units"]
    - comparison["actual_next_week_demand"]
)

print("Rows checked:", len(comparison))
print("Mismatches:", (comparison["difference"] != 0).sum())
print("Missing target values:",
      df_sorted["next_week_demand_target_units"].isna().sum())
```

```python
# STEP 5.23.1: Verify historical lag features

lag_columns = {
    "lag_1_week": 1,
    "lag_2_week": 2,
    "lag_3_week": 3,
    "lag_4_week": 4,
    "lag_8_week": 8,
    "lag_12_week": 12
}

for column, periods in lag_columns.items():

    expected = (
        df_tft.groupby("product_id")["units_sold"]
        .shift(periods)
    )

    valid = expected.notna()

    mismatches = (
        df_tft.loc[valid, column].to_numpy()
        != expected.loc[valid].to_numpy()
    ).sum()

    print(
        f"{column}: "
        f"{mismatches} mismatches "
        f"out of {valid.sum()} checked rows"
    )
```

```python
# STEP 5.23.2: Verify rolling means exclude current week

rolling_columns = {
    "rolling_mean_4_week": 4,
    "rolling_mean_8_week": 8,
    "rolling_mean_12_week": 12
}

for column, window in rolling_columns.items():

    expected = (
        df_tft.groupby("product_id")["units_sold"]
        .transform(
            lambda s: s.shift(1).rolling(
                window=window,
                min_periods=window
            ).mean()
        )
    )

    valid = expected.notna()

    mismatches = (
        ~np.isclose(
            df_tft.loc[valid, column].to_numpy(dtype=float),
            expected.loc[valid].to_numpy(dtype=float),
            rtol=1e-5,
            atol=1e-5
        )
    ).sum()

    print(
        f"{column}: "
        f"{mismatches} mismatches "
        f"out of {valid.sum()} checked rows"
    )
```

```python
# STEP 5.25.5: Find the actual shift of each stored lag

lag_columns = [
    "lag_1_week",
    "lag_2_week",
    "lag_3_week",
    "lag_4_week",
    "lag_8_week",
    "lag_12_week"
]

lag_results = []

for column in lag_columns:

    best_shift = None
    best_mismatches = None

    for shift in range(13):

        expected = (
            df_tft.groupby("product_id")["units_sold"]
            .shift(shift)
        )

        valid = expected.notna()

        mismatches = (
            df_tft.loc[valid, column].to_numpy()
            != expected.loc[valid].to_numpy()
        ).sum()

        if best_mismatches is None or mismatches < best_mismatches:
            best_mismatches = mismatches
            best_shift = shift

    lag_results.append({
        "stored_column": column,
        "best_actual_shift": best_shift,
        "mismatches": best_mismatches
    })

display(pd.DataFrame(lag_results))
```

```python
from pytorch_forecasting import TimeSeriesDataSet
from pytorch_forecasting.data import GroupNormalizer

# STEP 1: Prepare training data
training_data = df_tft_ready[
    df_tft_ready["time_idx"] <= 110
].copy()

# STEP 2: Create training dataset
training = TimeSeriesDataSet(
    training_data,
    time_idx="time_idx",
    target="units_sold",
    group_ids=["product_id"],

    min_encoder_length=15,
    max_encoder_length=15,

    min_prediction_length=1,
    max_prediction_length=1,

    static_categoricals=[
        "category_name",
        "brand"
    ],

    time_varying_known_reals=[
        "week_month",
        "week_quarter"
    ],

    time_varying_unknown_reals=[
        "units_sold",
        "temperature_c",
        "rainfall_mm"
    ],

    time_varying_unknown_categoricals=[
        "stockout_flag",
        "holiday_flag",
        "season",
        "weather_source__weather_missing"
    ],

    target_normalizer=GroupNormalizer(
        groups=["product_id"]
    ),

    add_relative_time_idx=True,
    add_target_scales=True,
    add_encoder_length=True
)

# STEP 3: Create validation dataset
validation = TimeSeriesDataSet.from_dataset(
    training,
    df_tft_ready,
    min_prediction_idx=111,
    stop_randomization=True
)

# STEP 4: Check results
print("Training samples:", len(training))
print("Validation samples:", len(validation))
print("Products:", training_data["product_id"].nunique())
```

```python
import torch
from pytorch_forecasting import TemporalFusionTransformer
from pytorch_forecasting.metrics import QuantileLoss

# Set random seed for reproducibility
torch.manual_seed(42)

# Create TFT model
tft = TemporalFusionTransformer.from_dataset(
    training,

    learning_rate=0.001,
    hidden_size=16,
    attention_head_size=2,
    dropout=0.1,
    hidden_continuous_size=8,

    loss=QuantileLoss(),
    optimizer="adam",

    reduce_on_plateau_patience=3
)

# Display model information
print("TFT model created successfully!")
print("Model parameters:", sum(
    p.numel() for p in tft.parameters()
))
```

```python
import lightning.pytorch as pl

from lightning.pytorch.callbacks import (
    EarlyStopping,
    LearningRateMonitor
)

# Set up trainer
early_stop_callback = EarlyStopping(
    monitor="val_loss",
    min_delta=0.001,
    patience=4,
    verbose=True,
    mode="min"
)

lr_logger = LearningRateMonitor()

trainer = pl.Trainer(
    max_epochs=20,
    accelerator="cpu",
    devices=1,
    gradient_clip_val=0.1,
    callbacks=[
        early_stop_callback,
        lr_logger
    ],
    enable_model_summary=True,
    log_every_n_steps=50
)

print("Trainer configured successfully!")
```
